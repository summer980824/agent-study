```python
import torch
from torch import nn

# 模型预测值
y_pred = torch.tensor(
    [2.5, 0.0, 2.0],
    requires_grad=True
)

# 真实值
y_true = torch.tensor([3.0, -0.5, 2.0])

loss_fn = nn.MSELoss()

loss = loss_fn(y_pred, y_true)

print("MSE Loss:", loss.item())

loss.backward()

print("Gradient:", y_pred.grad)
#为什么第三个预测值的梯度是 0？ 第三个值预测值与真实值一样，说明梯度0
#为什么预测值偏低时，梯度是负数？负数说明需要增加值来接近真实值
#为什么 MSE 会对较大的预测误差施加更大的惩罚？ 因为MSE对所有差值平方来去除正负值影响，但也加大了误差值
```

```python
logits = torch.tensor([
    [4.0, 1.0, 0.0],
    [0.0, 3.0, 1.0]
], requires_grad=True)

# 第一个样本属于类别 0
# 第二个样本属于类别 1
targets = torch.tensor([0, 1])

loss_fn = nn.CrossEntropyLoss()

loss = loss_fn(logits, targets)

print("Cross Entropy Loss:", loss.item())

loss.backward()

print("Gradient:", logits.grad)
```

```python
targets = torch.tensor([0, 1])

# 预测正确，而且置信度较高
confident_logits = torch.tensor([
    [4.0, 1.0, 0.0],
    [0.0, 3.0, 1.0]
])

# 预测也正确，但置信度较低
uncertain_logits = torch.tensor([
    [1.2, 1.0, 0.5],
    [0.9, 1.1, 0.8]
])

# 必须有2个target，但样本可以不同
wrong_logits = torch.tensor([
    [2.0, 3.0],
    [1.0, 2.0]
])

loss_fn = nn.CrossEntropyLoss()

confident_loss = loss_fn(confident_logits, targets)
uncertain_loss = loss_fn(uncertain_logits, targets)
wrong_loss = loss_fn(wrong_logits, targets)

print("Confident Loss:", confident_loss.item())
print("Uncertain Loss:", uncertain_loss.item())
print("Wrong Loss:", wrong_loss.item())

# 找到每个样本得分最高的类别
confident_pred = confident_logits.argmax(dim=1)
uncertain_pred = uncertain_logits.argmax(dim=1)
wrong_pred = wrong_logits.argmax(dim=1)

# 计算预测正确的比例
confident_acc = (confident_pred == targets).float().mean()
uncertain_acc = (uncertain_pred == targets).float().mean()
wrong_acc = (wrong_pred == targets).float().mean()

print("Confident Accuracy:", confident_acc.item())
print("Uncertain Accuracy:", uncertain_acc.item())
print("Wrong Accuracy:", wrong_acc.item())
```
