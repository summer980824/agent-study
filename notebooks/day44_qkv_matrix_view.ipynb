```python
import torch

X = torch.tensor([
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0],
])

W_q = torch.tensor([
    [1.0, 0.0],
    [1.0, 1.0],
])

W_k = torch.tensor([
    [1.0, 1.0],
    [0.0, 1.0],
])

W_v = torch.tensor([
    [1.0, 0.0],
    [0.0, 2.0],
])

Q = X @ W_q
K = X @ W_k
V = X @ W_v

print("Q:", Q)
print("K:", K)
print("V:", V)

print("Q shape:", Q.shape)
print("K shape:", K.shape)
print("V shape:", V.shape)
```

```python
scores = Q @ K.T

print(scores)
print(scores.shape)

expected_Q = torch.tensor([
    [1.0, 0.0],
    [1.0, 1.0],
    [2.0, 1.0],
])

expected_K = torch.tensor([
    [1.0, 1.0],
    [0.0, 1.0],
    [1.0, 2.0],
])

expected_V = torch.tensor([
    [1.0, 0.0],
    [0.0, 2.0],
    [1.0, 2.0],
])

expected_scores = torch.tensor([
    [1.0, 0.0, 1.0],
    [2.0, 1.0, 3.0],
    [3.0, 1.0, 4.0],
])

torch.testing.assert_close(Q, expected_Q)
torch.testing.assert_close(K, expected_K)
torch.testing.assert_close(V, expected_V)
torch.testing.assert_close(scores, expected_scores)

print("All matrix calculations passed!")
```
