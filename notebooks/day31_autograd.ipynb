```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2 + 3 * x + 1

y.backward()

print(x.grad)  # tensor(7.)

assert torch.allclose(
    x.grad,
    torch.tensor(7.0)
)
```

```python
x = torch.tensor(2.0, requires_grad=True)

y = x ** 2
y.backward()

print(x.grad)  # 4

y = x ** 3
y.backward()

print(x.grad)  # 16，而不是 12

x.grad.zero_()

y = x ** 3
y.backward()

print(x.grad)  # 12
```

```python
x = torch.tensor([-2., -1., 0., 1., 2.])
y_true = 2 * x + 1

w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

y_pred = w * x + b

loss = ((y_pred - y_true) ** 2).mean()
# tensor计算梯度
loss.backward()

print("w.grad:", w.grad) 
print("b.grad:", b.grad)

# 手动计算loss梯度
error = y_pred.detach() - y_true

manual_dw = 2 * (error * x).mean()
manual_db = 2 * error.mean()

print(manual_dw)  # -8
print(manual_db)  # -2

assert torch.allclose(w.grad, manual_dw)
assert torch.allclose(b.grad, manual_db)
```

```python
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

lr = 0.1

for step in range(100):
    # Forward
    y_pred = w * x + b

    # Calculate loss
    loss = ((y_pred - y_true) ** 2).mean()

    # Backward
    loss.backward()

    # Update parameters
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad

    # Clear gradients
    w.grad.zero_()
    b.grad.zero_()

with torch.no_grad():
    final_pred = w * x + b
    final_loss = ((final_pred - y_true) ** 2).mean()

print("Learned w:", w.item())
print("Learned b:", b.item())
print("Final loss:", final_loss.item())

assert abs(w.item() - 2.0) < 0.02
assert abs(b.item() - 1.0) < 0.02
```
