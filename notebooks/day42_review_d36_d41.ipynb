```python
import torch
from torch import nn

class MiniClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim, num_classes):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size, embedding_dim
        )
        self.norm = nn.LayerNorm(embedding_dim)
        self.dropout = nn.Dropout(0.2)
        self.classifier = nn.Linear(
            embedding_dim, num_classes
        )

    def forward(self, token_ids):
        x = self.embedding(token_ids)
        x = self.norm(x)
        x = x.mean(dim=1)
        x = self.dropout(x)
        return self.classifier(x)


model = MiniClassifier(
    vocab_size=100,
    embedding_dim=16,
    num_classes=2
)

token_ids = torch.randint(0, 100, (8, 5))
logits = model(token_ids)

assert logits.shape == (8, 2)
print(logits.shape)
```

```python
model.train()

out1 = model(token_ids)
out2 = model(token_ids)

print("Train outputs equal:",
      torch.allclose(out1, out2))

# 评估模式关闭Dropout随机丢弃
model.eval()

with torch.no_grad():
    out3 = model(token_ids)
    out4 = model(token_ids)

print("Eval outputs equal:",
      torch.allclose(out3, out4))
```
