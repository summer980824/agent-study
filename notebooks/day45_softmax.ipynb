```python
import torch


def softmax(x):
    exp_x = torch.exp(x)
    return exp_x / exp_x.sum(dim=-1, keepdim=True)


scores = torch.tensor([1.0, 2.0, 3.0])
# 每个e^x在sum中的比例
print(softmax(scores))

scores = torch.tensor([1000.0, 1001.0, 1002.0])
# e^1000 -> e^709.8 (float64)
print(softmax(scores))
```

```python
def stable_softmax(x):
    shifted = x - x.max(dim=-1, keepdim=True).values

    exp_x = torch.exp(shifted)

    return exp_x / exp_x.sum(
        dim=-1,
        keepdim=True
    )

scores = torch.tensor([1000.0, 1001.0, 1002.0])

result = stable_softmax(scores)

print(result)
print(result.sum())

assert torch.isfinite(result).all()
assert torch.allclose(
    result.sum(),
    torch.tensor(1.0)
)
```

```python
expected = torch.softmax(scores, dim=-1)

torch.testing.assert_close(result, expected)

print("All tests passed!")
```

```python
scores = torch.tensor([
    [1.0, 2.0, 3.0],
    [2000.0, 2001.0, 2004.0],
])

print(stable_softmax(scores))
```

```python
scores = torch.randn(2, 3, 4)
attn = softmax(scores)

print(scores)
print(attn)
print(attn.sum(dim=-1))
```
