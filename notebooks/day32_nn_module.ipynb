```python
import torch
import torch.nn as nn

class LinearModel(nn.Module):

    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.linear(x)


model = LinearModel()

x = torch.tensor([
    [-2.],
    [-1.],
    [ 0.],
    [ 1.],
    [ 2.]
])

# 调用 model(x) 会经过 nn.Module 的 __call__ 机制，由它负责调用 forward()，并处理 PyTorch 的模块调用机制及相关 hooks。
y_pred = model(x) 

print(y_pred)
print(y_pred.shape)

for name, param in model.named_parameters():
    print(name)
    print(param)
    print(param.shape)
    print(param.requires_grad)

total_params = sum(
    p.numel() for p in model.parameters()
)

print(total_params)
```

```python
with torch.no_grad():
    model.linear.weight.fill_(2.0)
    model.linear.bias.fill_(1.0)

y_pred = model(x)

y_true = torch.tensor([
    [-3.],
    [-1.],
    [ 1.],
    [ 3.],
    [ 5.]
])

print(y_pred)

assert torch.allclose(y_pred, y_true)
```
