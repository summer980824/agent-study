# Day 47: PyTorch 项目结构

使用 `agent_study/pytorch` 包中的模块（`data.build_toy_dataset`、`model.TinyTextClassifier`、`train.train_model`），重新实现 day46_mini_neural_network.ipynb 的训练实验。

```python
import torch
from torch import nn

from agent_study.pytorch.data import build_toy_dataset
from agent_study.pytorch.model import TinyTextClassifier
from agent_study.pytorch.train import train_model

torch.manual_seed(42)
vocab, X_train, y_train, X_test, y_test = build_toy_dataset()
```

```python
model = TinyTextClassifier(
    vocab_size=len(vocab),
    embedding_dim=8,
    num_classes=2,
)
```

```python
with torch.no_grad():
    embedded = model.embedding(X_train)
    pooled = embedded.mean(dim=1)
    logits = model.classifier(pooled)

    print('Embedding:', embedded.shape)
    print('Pooling:', pooled.shape)
    print('Logits:', logits.shape)
```

```python
loss_history = train_model(model, X_train, y_train, epochs=101, learning_rate=0.1)

for epoch in range(0, len(loss_history), 20):
    print(f'Epoch {epoch}, Loss: {loss_history[epoch]:.6f}')

loss_fn = nn.CrossEntropyLoss()
initial_loss = loss_history[0]
with torch.no_grad():
    final_loss = loss_fn(model(X_train), y_train).item()

print('Initial loss:', initial_loss)
print('Final loss:', final_loss)

assert final_loss < initial_loss
```

```python
model.eval()

with torch.inference_mode():
    train_pred = model(X_train).argmax(dim=1)
    test_pred = model(X_test).argmax(dim=1)

    print('Train predictions:', train_pred)
    print('Train labels:', y_train)
    print('Test predictions:', test_pred)
    print('Test labels:', y_test)

    train_accuracy = (train_pred == y_train).float().mean()
    test_accuracy = (test_pred == y_test).float().mean()

    print('Train accuracy:', train_accuracy.item())
    print('Test accuracy:', test_accuracy.item())
```
