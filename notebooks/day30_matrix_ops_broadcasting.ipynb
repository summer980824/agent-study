```python
import torch

a = torch.tensor([
    [1., 2., 3.],
    [4., 5., 6.]
])

b = torch.tensor([
    [1., 2.],
    [3., 4.],
    [5., 6.]
])

# Matrix multiplication
c = a @ b

print(a.shape)  # (2, 3)
print(b.shape)  # (3, 2)
print(c.shape)  # (2, 2)
print(c)

d = a * a
print(d)
```

```python
x = torch.tensor([
    [1., 2., 3.],
    [4., 5., 6.]
])

bias = torch.tensor([10., 20., 30.])

result = x + bias

print(result)
print(result.shape)

column = torch.tensor([
    [100.],
    [200.]
])

result = x + column
print(result)

invalid = torch.tensor([10., 20.])
print(invalid.shape)
try:
    result = x + invalid
except RuntimeError as e:
    print(type(e).__name__)
    print(e)

assert c.shape == (2, 2)
assert (x + bias).shape == (2, 3)
assert (x + column).shape == (2, 3)
```
