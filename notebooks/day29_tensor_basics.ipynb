```python
import torch
import numpy as np

torch.manual_seed(42)

# Scalar: 0-dimensional
scalar = torch.tensor(3.14)

# Vector: 1-dimensional
vector = torch.tensor([1, 2, 3])

# Matrix: 2-dimensional
matrix = torch.tensor(
    [[1, 2, 3], [4, 5, 6]],
    dtype=torch.float32
)

# Common constructors
zeros = torch.zeros(2, 3)
ones = torch.ones(2, 3)
sequence = torch.arange(0, 10)
random_tensor = torch.randn(2, 3)

print(matrix.shape)
print(matrix.dtype)
print(matrix.device)
print(matrix.ndim)
print(matrix.numel())
print(sequence)
print(random_tensor)
```

```python
arr = np.arange(6, dtype=np.float32).reshape(2, 3)

# Shares memory with NumPy array
tensor_shared = torch.from_numpy(arr)

# Creates a copy
tensor_copy = torch.tensor(arr)

print(tensor_shared)
print(tensor_copy)
print(tensor_shared.numpy())

arr[0, 1] = 100
print(tensor_shared)
print(tensor_copy)
```

```python
# Tensor indexing and modification
x = torch.arange(6).reshape(2, 3)

x[0, 1] = 99

print(x)

# dtype conversion
x_float = x.to(torch.float32)

print(x.dtype)
print(x_float.dtype)
```
